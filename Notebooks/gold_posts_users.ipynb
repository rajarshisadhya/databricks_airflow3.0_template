## Create a one big table

```python
from pyspark.sql import DataFrame
import pyspark.sql.functions as F

def post_users_OBT(stg_posts_df: DataFrame, raw_users_df: DataFrame) -> DataFrame:
    return(
        stg_posts_df.alias("posts")
        .withColumnRenamed("CreationDate", "PostCreationDate")
        .join(
            other=raw_users_df.withColumnRenamed("CreationDate","UserCreationDate")
            .alias("users"),
            on=F.col("posts.OwnerUserId") == F.col("users.Id"),
            how="left"

        )
    )


marts_posts_users_df = post_users_OBT(stg_posts_df, raw_users_df)
display(marts_posts_users_df.limit(10))
```

```python
marts_posts_users_df.write.mode("overwrite").saveAsTable("rsadhya_data_platform_tutorial.gold.marts_posts_users")
```
