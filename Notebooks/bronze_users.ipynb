## Creating Schema

```python
from pyspark.sql.types import StructType, StructField, StringType, LongType, TimestampType

users_schema = StructType([
    StructField("_AboutMe", StringType(), True),
    StructField("_AccountId", LongType(), True),
    StructField("_CreationDate", TimestampType(), True),
    StructField("_DisplayName", StringType(), True),
    StructField("_DownVotes", LongType(), True),
    StructField("_Id", LongType(), True),
    StructField("_LastAccessDate", TimestampType(), True),
    StructField("_Location", StringType(), True),
    StructField("_Reputation", LongType(), True),
    StructField("_UpVotes", LongType(), True),
    StructField("_Views", LongType(), True),
    StructField("_WebsiteUrl", StringType(), True),
])

# Use it when reading the XML file instead of inferSchema:
# spark.read.format("xml").option("rootTag", "users").option("rowTag", "row").schema(users_schema).load(...)
```

## Reading data from XML files

```python
df = (
    spark.read.format("xml")
    .option("rootTag","users")
    .option("rowTag","row")
    .schema(users_schema)
    .option("header","true")
    .load("/Volumes/rsadhya_data_platform_tutorial/default/my_volume/Users.xml")
)

display(df.limit(10))
```

## Removing the _ from the column names

```python
new_col_names = [col[1:] for col in df.columns]
df = df.toDF(*new_col_names)
display(df.limit(10))
```

## Writing the data in Managed Tables

```python
df.write.mode("overwrite")\
    .option("overwriteSchema","true")\
    .saveAsTable("rsadhya_data_platform_tutorial.bronze.raw_users")
```
